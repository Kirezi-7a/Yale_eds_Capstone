<span style='font-size:medium'>**About the project:** </span>This capstone project analyses thematic differences in how carbon credits and community impacts are discussed in official project documentation compared to independent reporting. The dataset consists of documents identified by source, year, document ID, and document name, with the primary analytical focus on the unstructured text. Using structural topic modelling, we aim to identify recurring themes within each source and examine how these topics differ across document types. The results will be visualised to highlight potential differences or biases in how official and independent sources frame these issues.
